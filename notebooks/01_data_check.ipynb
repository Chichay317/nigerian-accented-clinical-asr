```python
from datasets import load_dataset

ds = load_dataset("intronhealth/afrispeech-200", "yoruba",
                  split="train", streaming=True, trust_remote_code=True)

sample = next(iter(ds))
print({k: v for k, v in sample.items() if k != "audio"})
print("Sampling rate:", sample["audio"]["sampling_rate"])
print("Audio length (seconds):", len(sample["audio"]["array"]) / sample["audio"]["sampling_rate"])
```
